```python
# Work around Tcl/Tk path detection in Windows Python 3.13.0 virtual environments.
import os
import sys
from pathlib import Path

if sys.platform == "win32" and sys.version_info[:3] == (3, 13, 0):
    tcl_root = Path(sys.base_prefix) / "tcl"
    tcl_file = next(tcl_root.glob("tcl*/init.tcl"), None)
    tk_file = next(tcl_root.glob("tk*/tk.tcl"), None)
    if tcl_file is not None and tk_file is not None:
        os.environ["TCL_LIBRARY"] = str(tcl_file.parent)
        os.environ["TK_LIBRARY"] = str(tk_file.parent)
    else:
        print("Tcl/Tk files are missing. Add or repair Tcl/Tk and IDLE using the Python installer.")
```
