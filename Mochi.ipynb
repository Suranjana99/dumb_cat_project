```python
import os
import random
import tkinter as tk
from PIL import Image, ImageTk
```

```python

BG = "#ffe4ec"
TEXT = "#5a3e4b"
BUTTON = "#ffb6c9"
```

```python
# Load your kitten photo.
img_path = r"C:\Users\SURANJANA\OneDrive\Desktop\Mochi_pet\cat.png"

with Image.open(img_path) as source:
    img = source.convert("RGBA")

img.thumbnail((200, 200))
```

```python
window = tk.Tk()
window.title("Mochi my baby kitten <3")
window.geometry("450x650")
window.configure(bg=BG)
```

```python
tk.Label(
    window,
    text="Meet Mochi : My baby kitten <3",
    font=("Arial", 20, "bold"),
    bg=BG,
    fg=TEXT,
    wraplength=440
).pack(pady=(20, 10))

# The photo and its label have separate names.
cat_photo = ImageTk.PhotoImage(img, master=window)

cat_label = tk.Label(
    window,
    image=cat_photo,
    text="🐱",
    compound="top",  # Photo above, emoji underneath.
    font=("Arial", 24),
    bg=BG,
    fg=TEXT
)
cat_label.image = cat_photo
cat_label.pack(pady=(10, 20))
```

```python
mood = tk.Label(
    window,
    text="Mood: ready to play",
    font=("Arial", 16),
    bg=BG,
    fg=TEXT
)
mood.pack(pady=(10, 20))

message = tk.Label(
    window,
    text="Hiiii, I am Mochi, your baby kitten. I like catnip <3 and cuddles :)",
    font=("Arial", 12),
    bg=BG,
    fg=TEXT,
    wraplength=400,
    justify="center",
    height=3
)
message.pack(pady=(10, 20))


def feed():
    cat_label.config(text="😻")
    mood.config(text="Mood: happy and full")
    message.config(text=random.choice([
        "Yummy! You're my favourite human.",
        "A snack? For me? Best day ever!",
        "Purr... that was delicious!"
    ]))


def play():
    cat_label.config(text="😸")
    mood.config(text="Mood: playful")
    message.config(text=random.choice([
        "You throw the yarn. I'll chase it!",
        "Zoomies activated!",
        "I caught it! Did you see that?"
    ]))


def nap():
    cat_label.config(text="😴")
    mood.config(text="Mood: sleepy")
    message.config(text=random.choice([
        "Five more minutes...",
        "Dreaming about snacks...",
        "Goodnight, favourite human."
    ]))


button_frame = tk.Frame(window, bg=BG)
button_frame.pack(pady=15)

for label, colour, action in [
    ("Feed", "#FFD6A5", feed),
    ("Play", "#CAEDDA", play),
    ("Sleep", "#D9D2F3", nap),
]:
    tk.Button(
        button_frame,
        text=label,
        command=action,
        font=("Arial", 12),
        bg=colour,
        fg=TEXT,
        width=7,
        relief="flat",
        cursor="hand2",
        pady=8
    ).pack(side="left", padx=6)

window.mainloop()
```
